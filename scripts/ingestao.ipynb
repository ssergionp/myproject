# LANDING

```python
   import duckdb
   import pandas as pd
   import os
   from datetime import datetime
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
con.execute("""DROP TABLE IF EXISTS bronze_produtos""")
```

```python
arquivo  = 'z0019_1.csv'
data_ingestao = datetime.now()
df1 = pd.read_csv(f'../landing/{arquivo}', sep=';')
df1['nome_arquivo'] = arquivo
df1['data_ingestao'] = data_ingestao
df1.head(20)
```

```python
arquivo  = 'z0019_2.csv'
data_ingestao = datetime.now()
df2 = pd.read_csv(f'../landing/{arquivo}', sep=';')
df2['nome_arquivo'] = arquivo
df2['data_ingestao'] = data_ingestao
df2.head(20)
```

# BRONZE

```python
# bronze_produtos

con.execute("""
   CREATE TABLE IF NOT EXISTS bronze_produtos(
      NATBR VARCHAR,
      MAKTX VARCHAR,
      WERKS VARCHAR,
      MAINS VARCHAR,
      LABST VARCHAR,
      nome_arquivo VARCHAR,
      data_ingestao TIMESTAMP
   );
""")
```

```python
# Limpa a tabela para não acumular lixo de execuções anteriores
con.execute("DELETE FROM bronze_produtos")

# Insere o primeiro lote
con.execute("INSERT INTO bronze_produtos SELECT * FROM df1")

# Insere o segundo lote
con.execute("INSERT INTO bronze_produtos SELECT * FROM df2")

# Consulta para validar (ordenando para ver os novos no topo ou fim)
resultado = con.execute("SELECT * FROM bronze_produtos").df()
display(resultado)
```
